# Loading Libraries and CSV File

Now that we know the problems, we will load some libraries to get the tools required for the analysis. We will use Pandas, matplotlib.pyplot, and Seaborn to solve our problems.

```python
# Importing libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

We also load the CSV file from the dataset "Sleep_health_and_lifestyle_dataset.csv," where it has the data we need, and put it in the sleepData variable to save room for future coding.

```python
# Importing the CSV file
sleepData = pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")
```

# Data Cleaning

There are 219 missing values in the sleep disorder column, even though Kaggle said there are no missing values. In Kaggle, it was named "None" instead of NaN.

```python
# Check for missing values
sleepData.isnull().sum()
```

To fix this, we get the sleep disorder column, convert the N/A to None, and turn it into a category. After that, we checked the value counts, and we saw that there were 219 "None" values instead of NaN.

```python
# Converting N/A to "None," and turn it into a category
sleepData['Sleep Disorder'] = sleepData['Sleep Disorder'].fillna('None').astype('category')

# Check how many values for each category
sleepData['Sleep Disorder'].value_counts(dropna = False)
```

We checked if there are missing values for the sleep disorder column, and there are none.

```python
# Check if there are any missing values after converting
sleepData['Sleep Disorder'].isnull().sum()
```

# Physical Activity as a Moderator

We have finished cleaning the data. Let's start with the first problem. We want to know if higher physical activity buffers the negative impact of stress on self-reported sleep quality.

First, we create a column called "Activity Group" and use the Physical Activity Level column to split into four bins to make three groups containing minutes per day.

Then, we check how many values are in each category.

```python
# Categorize physical activity levels (Ranges from 30 to 90)
sleepData['Activity Group'] = pd.cut(
    sleepData['Physical Activity Level'], bins = [30, 50, 70, 90],  # Split into four bins
    labels = ['Low (30-50)', 'Medium (51-70)', 'High (71-90)']) # Split into three groups

# Check how many values for each group
sleepData['Activity Group'].value_counts()
```

Next, we create a scatterplot with regression lines. After running the code, we get the lines and scatters from the activity group.

```python
# Scatterplot with regression line with activity group, stress level, and quality of sleep
sns.lmplot(
    data = sleepData, x = 'Stress Level', y = 'Quality of Sleep', hue = 'Activity Group', palette = 'bright')

# Physical Activity Buffers Stress Impact on Sleep Quality scatterplot with regression line plot
plt.title("Physical Activity Buffers Stress Impact on Sleep Quality")
plt.xlabel("Stress Level (1-10)")
plt.ylabel("Sleep Quality (1-10)")
plt.show()
```

**Insights:** 
- High physical activity group has better sleep quality than the low group at the lowest stress level
- Medium group has better sleep quality than the low group at the highest stress level

In conclusion, exercise recommendations should match stress levels.

# Age and Sleep Patterns (Sleep Duration)

Now the first problem is over, let's move to the second one. We want to know how sleep duration and disorder prevalence vary across age groups.

First, we create a column called "Age Group" and use the Age column to split into four bins to create three age groups.

Then, we check how many values are in each group.

```python
# Categorize age (Ranges from ages 27 to 59)
sleepData['Age Group'] = pd.cut(
    sleepData['Age'], bins = [27, 35, 45, 59], labels = ['27-35', '36-45', '46-59']) # Split into four bins and three groups

# Check how many values for each group
sleepData['Age Group'].value_counts()
```

Next, we create a boxplot, and when we run the code, we see three age groups and the white dot representing the mean.

```python
# Boxplot with age group and sleep duration
sns.boxplot(
    data = sleepData, x = 'Age Group', y = 'Sleep Duration', hue = 'Age Group', palette = 'Blues', showmeans = True, 
    meanprops = {'marker':'o', 'markerfacecolor':'white'}, legend = False)

# Average Sleep Duration by Age Group boxplot
plt.title("Average Sleep Duration by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Hours")
plt.show()
```

**Insights:** 
- Ages 46-59 have the biggest range, highest sleep amount, median, and mean compared to the others
- Ages 36-45 have the shortest range
- Ages 27-35 have the lowest mean and sleep amount

Therefore, the youngest age group tends to sleep less due to busy lives.

# Age and Sleep Patterns (Sleep Disorder)

Moving to sleep disorders, we create a disorder age variable to combine the age group and sleep disorder columns and later organize the categories in order.

```python
# Categorize sleep disorder with age group and sleep disorder
disorderAge = pd.crosstab(
    sleepData['Age Group'], sleepData['Sleep Disorder'], normalize = 'index').mul(100) # Making the y-axis up to 100%

# Organizing categories in order
disorderAge = disorderAge[['None', 'Insomnia', 'Sleep Apnea']]
```

Then, we create a stacked barplot, and when we run the code, we see three different categories in each age group.

```python
# Making the plot a barplot and stacked
disorderAge.plot(kind = 'bar', stacked = True)

# Sleep Disorder Prevalence by Age stacked barplot with legend
plt.title("Sleep Disorder Prevalence by Age Group (%)")
plt.xlabel("Age Group")
plt.ylabel("Percentage")
plt.xticks(rotation = 0)
plt.legend(title = 'Disorder', bbox_to_anchor = (1, 1))
plt.show()
```

**Insights:** 
- "None" category has a large percentage of ages 27-35 and 36-45, while the "Sleep Apnea" category has a large percentage of ages 46-59
- "Insomnia" category has the lowest percentage of ages 27-35 and 46-59 compared to ages 36-45, whose category with the smaller percentage is "Sleep Apnea"

Therefore, the oldest age group tends to have a higher sleep disorder prevalence, particularly sleep apnea, than the youngest age group.

# Conclusion
**Physical Activity Protects Sleep Under Stress:** 
- High physical activity (> 70 mins/day) correlates with better sleep quality at low stress levels, while medium activity (51-70 mins/day) is most protective under chronic high stress
- **Implication:** Tailored exercise programs (e.g., high-intensity for general wellness, moderate-intensity for high-stress jobs) could improve sleep health

**Age Dramatically Impacts Sleep Patterns:** 
- Young Adults (Ages 27-35) has the shortest sleep duration likely due to lifestyle demands but lowest disorder rates
- Middle-Aged (Ages 46-59) has the longest sleep duration but highest sleep apnea prevalence, which is a problem for age-related health risks.
- **Implication:** Target sleep apnea screening for adults over 45 and stress management for younger groups

Sleep health is modifiable through lifestyle choices. By addressing stress through activity and age-specific risks, we can design better interventions, from workplace policies to clinical screenings.
